#GC content

The formula chosen is the formula for counting the fraction of GC content in a DNA sequence.

The value is computed as $\dfrac{1}{L}\sum_{i=1}^{L} \mathbb{1}[\,b_i \in \{G, C\}\,]$

```python
# the function as a loop is as follows
def GC_loop(sequence):
  count = 0
  for base in sequence:
    count += 1*(str(base).upper() == 'G' or str(base).upper() == 'C')

  return count/len(sequence)
```

```python
# meanwhile by using numpy it becomes
import numpy as np

def GC_numpy(sequence):
  sequence_array = np.array(list(sequence))
  return np.count_nonzero(np.isin(sequence_array, ['C','G']))/sequence_array.size
```

```python
# the answer is the same
sequence = "ACTGACACGACCACGTTGTCGAGTCGAGAGCTTTGCGCGC"
np_val=GC_numpy(sequence)
loop_val=GC_loop(sequence)
assert ( np_val == loop_val)
print(f'GC content with the loop function: \t{loop_val} \nGC content with the numpy arrays: \t{np_val}')
```
